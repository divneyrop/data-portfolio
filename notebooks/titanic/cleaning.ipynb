## Handle missing data and duplicates
-------

```python
import pandas as pd

df = pd.read_csv("../../datasets/raw/titanic_train.csv")

print(df.shape)
```

- **Check for Duplicates**

```python
duplicates = df.duplicated().sum()
print(f"Number of duplicate rows: {duplicates}")
```

- **Fill missing Age**

```python
median_age = df.groupby(['Pclass', 'Sex'])['Age'].transform('median')
df['Age'] = df['Age'].fillna(median_age)
df.Age.isnull().sum()
```

Missing values imputed using median age grouped by the **Pclass** and **Sex**, which provide a more estimative representation., compared to overall median.

- Fill missing Embarkation. 
 **Inspect missing data first**

```python
print(
    df.loc[df["Embarked"].isnull()]
)
```

```python
print(
    df.groupby(["Pclass", "Sex"])["Embarked"].value_counts()
)
```

Filling missing embarked data with mode grouped by **Pclass** and **Sex**.

```python
df["Embarked"] = df["Embarked"].fillna(df.groupby(["Pclass", "Sex"])["Embarked"].transform(lambda x: x.mode()[0]))
print(df.isnull().sum())
```

- Drop Cabin

```python
df.drop(columns=["Cabin"], inplace=True)
```

```python
df.to_csv("../../datasets/processed/titanic_cleaned.csv", index=False)
```
