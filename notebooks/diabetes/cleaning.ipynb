* Inspect data

```python
import pandas as pd

df = pd.read_csv("../../datasets/raw/diabetes.csv")
print(df.shape)
```

```python
print(df.info())
```

*NO missing data.*

```python
df.isna().sum()
```

* Check for duplicates

```python
# (Unnecessary) Check for duplicates in each column.

duplicates = df.apply(lambda col: col.duplicated().sum())
print(duplicates)
```

```python
# Check for duplicated rows

duplicates = df.duplicated().sum()
print(f"Number of duplicate rows: {duplicates}")
```

```python
# Inspect duplicated data

duplicates = df[df.duplicated(keep=False)]
duplicates.sort_values(by=df.columns.tolist()).head(20)
```

```python
# Check duplicates counts

duplicate_counts = df.value_counts()
print(duplicate_counts[duplicate_counts > 1].value_counts())
```

* Drop duplicates

```python
proc_df = df.drop_duplicates()
print(proc_df.duplicated().sum())
```

*The dataset contains exact duplicate rows. Since no unique patient or visit identifier is provided, it cannot be conclusively determined whether these represent separate individuals with identical characteristics or duplicated records. The duplicates were therefore examined and treated as redundant observations for the purposes of this analysis.*

* Exclude category 'Others' in gender

```python
proc_df["gender"].value_counts()
```

```python
clean_df = proc_df[proc_df["gender"] != "Other"]
clean_df["gender"].value_counts()
```

*The dataset contained 18 observations classified as "Other" under gender. Since the analysis focuses on the Female and Male categories and the "Other" category represents only 0.018% of observations, these observations were excluded from the gender-based analysis.*

* Save cleaned dataset

```python
clean_df.to_csv("../../datasets/processed/diabetes_cleaned.csv", index=False)
print("Cleanned dataset saved")
```
